```python
df1['Date']=pd.to_datetime(df1['Date'])
```

```python
df1.info()
```

```python
df1.shape
```

```python
df1.isnull().sum()
```

```python
# handle missing value
#filling temp coloumn with next day values
df1['Temperature']=df1['Temperature'].bfill()
```

```python
df1
```

```python
#filling dewpoint coloumn with previous day values
df1['DewPoint']=df1['DewPoint'].ffill()
```

```python
df1
```

```python
df1['Humidity']=df1['Humidity'].fillna(df1['Humidity'].mean())
```

```python
df1
```

```python
df1['Sea Level PressureIn']=df1['Sea Level PressureIn'].fillna(df1['Sea Level PressureIn'].median())
```

```python
df1
```

```python
df1['VisibilityMiles']=df1['VisibilityMiles'].ffill()
```

```python
df1['WindSpeedMPH']=df1['WindSpeedMPH'].bfill()
```

```python
df1
```

```python
df1['Event']=df1['Event'].fillna(df1['Event'].mode()[0])
```

```python
df1
```

```python
df1.to_csv('/Users/arjunaji/BIA ARJUN/DataPandas/ny_weather_cleaned_data.csv',index=False)
```

```python
iris_data=pd.read_csv('/Users/arjunaji/BIA ARJUN/iris.csv')
```

```python
iris_data
```

```python
#scatter plot
plt.scatter(iris_data['sepal.length'],iris_data['sepal.width'])
plt.xlabel('Sepal Length(cm)')
plt.ylabel('Sepal Width(cm)')
plt.title('Sepal Length vs Sepal Width')
```

```python
#histogram
plt.hist(iris_data['sepal.length'])
plt.show
```

```python
species_count= iris_data['variety'].value_counts()
species_count
```

```python
plt.bar(species_count.index,species_count.values,color=['red','blue','green'])
plt.xlabel('Species')
plt.ylabel('Count')
plt.title('Count of Different Species')
```

```python
import seaborn as sns
sns.histplot(data=iris_data,x='sepal.length',color='green')
plt.xlabel('Sepal Length(cm)')
plt.ylabel('Count')
plt.title('Distribution of Sepal Length')
plt.show()
```

```python
sns.boxplot(data=iris_data,y="sepal.length",x="variety",color="brown")
```

```python
sns.pairplot(iris_data,hue='variety')
```
