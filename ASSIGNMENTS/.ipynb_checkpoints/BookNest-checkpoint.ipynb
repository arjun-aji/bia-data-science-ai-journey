```python
catalog = []

def add_book(catalog, title, author, price, category, rating=3.0):
    book = {
        "title": title,
        "author": author,
        "price": price,
        "category": category,
        "rating": rating
    }
    catalog.append(book)
```

```python
add_book(catalog, "The Alchemist", "Paulo Coelho", 399, "Fiction", 4.5)
add_book(catalog, "Atomic Habits", "James Clear", 550, "Self-Help", 4.8)
add_book(catalog, "Python Crash Course", "Eric Matthes", 699, "Academic", 4.7)
add_book(catalog, "Ikigai", "Hector Garcia", 299, "Self-Help")
add_book(catalog, "Rich Dad Poor Dad", "Robert Kiyosaki", 799, "Finance", 4.4)
```

```python
def calculate_total(*prices):
    return sum(prices)
def display_book_profile(title, author, **details):
    print("Title:", title)
    print("Author:", author)

    for key, value in details.items():
        print(key, ":", value)


def sort_books(catalog):
    return sorted(catalog, key=lambda book: book["price"])


def premium_books(catalog, price):
    return list(filter(lambda book: book["price"] > price, catalog))


def make_discount_calculator(percent):

    def discount(price):
        return price - (price * percent / 100)

    return discount
```

```python
print("BOOKNEST")
print("-" * 40)

print("FULL CATALOG")
print("-" * 40)

for book in catalog:
    print(book)


print()
print("SORTED BOOKS")
print("-" * 40)

books = sort_books(catalog)

for book in books:
    print(book["title"], "-", book["price"])


print()
print("PREMIUM BOOKS")
print("-" * 40)

books = premium_books(catalog, 500)

for book in books:
    print(book["title"], "-", book["price"])
print()
print("CHECKOUT TOTALS")
print("-" * 40)

print(calculate_total(399))
print(calculate_total(399, 550, 299))


print()
print("BOOK PROFILE")
print("-" * 40)

display_book_profile(
    "Python Crash Course",
    "Eric Matthes",
    edition="3rd",
    publisher="No Starch Press",
    language="English"
)


print()
print("20% DISCOUNT")
print("-" * 40)

discount = make_discount_calculator(20)

for book in catalog:
    new_price = discount(book["price"])
    print(book["title"], "-", new_price)
```
