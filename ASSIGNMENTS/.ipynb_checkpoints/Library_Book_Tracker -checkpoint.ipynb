## Library Book Tracker — Problem Statement
Mr. Sharma is the librarian at Sunrise College. His library has three sections
that students love the most: **Python Books**, **Database Books**, and
**Web Development Books**.
Many students borrow books from more than one section, and Mr. Sharma has one
strict rule: **no student's name should be written twice in the same
section's list**, even if they borrow from it many times. To keep things
simple and organized, he wants a program to help him manage everything.

**Your task:** Write a Python program that helps Mr. Sharma do all of the
above, using a data structure where every student's name appears only once,
and that makes it easy to compare which students belong to which sections.

use set() not list

Here is what Mr. Sharma needs, one task at a time:

- 1. **Start of the Year** — Create three empty lists: one each for Python,
   Database, and Web Development, to note down students who borrow books.

```python
python = set()
database = set()
web_development = set()
```

- 2. **First Student Arrives** — Ravi borrows a Python book. Add his name to
   the Python list.

```python
python.add("Ravi")
```

```python
python
```

```python
python.update(["Arjun", "Rahul", "Anu", "Vishnu", "Meera"])
web_development.update(["Arjun", "Rahul", "Sneha", "Kiran", "Anu"])
```

- 3. **A Group Comes Together** — Five students from Class 12-A visit together
   and all borrow Database books. Add all their names to the Database list
   at once.

```python
database.update(["Aman", "Rahul", "Anu", "Simran", "Arjun"])
database
```

- 4. **Checking a Name** — A student asks, "Am I already on the Web Development
   list?" Write code to check if their name is already there.

```python
name = input("enter name:")

if name in web_development:
    print(name, "is already in the Web Development list.")
else:
    print(name, "is not in the Web Development list.")
```

- 5. **A Book Is Returned** — Simran returns her Python book. Remove her name
   from the Python list. (If her name isn't there, it's fine to show an
   error — that helps catch mistakes.)

```python
python.add("Simran")
```

```python
python
```

```python
python.remove("Simran")
```

```python
python
```

- 6. **Maybe Returned, Maybe Not** — Mr. Sharma isn't sure if Aman's name is in
   the Database list, but he wants to remove it if it *is* there — without
   the program crashing if it's *not* there.

```python
database.discard("Aman")
```

```python
database
```

- 7. **Surprise Check** — For a quick surprise check, Mr. Sharma wants to pick
   any one random student from the Web Development list and verify their
   record.

```python
student = web_development.pop()
print("Selected student:", student)
```

- 8. **Making a Safety Copy** — Before making changes to the Python list, Mr.
   Sharma wants to save a separate backup copy of it, just in case.

```python
python_backup = python.copy()
```

- 9. **Printing the Lists** — Mr. Sharma wants to print out all the students in
   each section, one by one, to put up on the notice board.

```python
print("\nPython Students:")
for name in python:
    print(name)

print("\nDatabase Students:")
for name in database:
    print(name)

print("\nWeb Development Students:")
for name in web_development:
    print(name)
```

- 10. **Students in Two Sections** — Find students who have borrowed books
    from **both** Python and Database sections.

```python
python_database = python.intersection(database)
print("\nStudents in Python and Database:", python_database)
```

- 11. **All Students Combined** — Find the list of **all unique students** who
    have borrowed a book from **any** of the three sections (no repeated
    names).

```python
all_students = python.union(database, web_development)
print("All Students:", all_students)
```

- 12. **Only Python, Not Database** — Find students who borrowed a Python book
    but have **never** borrowed a Database book.

```python
python_only = python.difference(database)
print("Python but not Database:", python_only)
```

- 13. **Best Readers** — Find the students who have borrowed books from **all
    three** sections — Python, Database, and Web Development.


```python
best_readers = python.intersection(database, web_development)
print("Students in all three:", best_readers)
```

- 14. **Combining Two Sections** — The Web Development section is closing
    temporarily. Add all its students into the Database list too (without
    removing them from Web Development).

```python
database.update(web_development)
database
```

- 15. **Fresh Start** — At the end of the year, all Web Development books have
    been returned. Clear that entire list so it's empty for next year.


```python
web_development.clear()

print("Web Development after clearing:", web_development)
```
