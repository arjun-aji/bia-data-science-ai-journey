## Employee Information Management System

A company wants to maintain the details of its employees using a Python program. The HR department currently stores employee information manually, making it difficult to add new employees, update employee details, search for information, and analyze employee records.

You are asked to develop a Python Employee Information Management System using a Dictionary.

The system should store each employee's information using the employee name as the key and employee details as the value.


- 1. Create a nested employee dictionary containing multiple employees.

```python
employee = {"Arjun": {"age": 21,"department": "CSE","salary": 45000,"rating": 8},
             "Dhron": {"age": 22,"department": "IT","salary": 25000,"rating": 7},
            "Jaijith": {"age": 22,"department": "CSE","salary": 35000,"rating": 6},
            "Aadith": {"age": 22,"department": "IT","salary": 30000,"rating": 4},
            "Abhi": {"age": 19,"department": "CSE","salary": 39000,"rating": 8},
            "Aadithyan": {"age": 21,"department": "IT","salary": 20000,"rating": 4}
           }
```

- 13. Find the highest salary, average salary, employees earning above ₹50,000 and employees earning below ₹30,000.

```python
highest_sal = 0
highest_employee = ""
total_sal=0 
n=0
above_50000 = []
below_30000 = []
for name, details in employee.items():
    total_sal+=details["salary"]
    n+=1
    if details["salary"] > highest_sal:
        highest_sal = details["salary"]
        highest_employee = name

    if details["salary"]>50000:
        above_50000.append(name)

    if details["salary"]<30000:
        below_30000.append(name)
avg_sal=total_sal/n
print("Highest salary:", highest_sal, ",Employee:", highest_employee)
print("Average salary:", avg_sal)
print("Employee earning above ₹50,000: ",above_50000)
print("Employee earning below ₹30,000: ",below_30000)
```

- 14. Find the highest performance rating, average rating, employees with rating above 8, and sort employees by rating.

```python
highest_rating = 0
highest_rate_employee = ""
total_rating=0 
n=0
above_8 = []
for name, details in employee.items():
    total_rating+=details["rating"]
    n+=1
    if details["rating"] > highest_rating:
        highest_rating = details["rating"]
        highest_rate_employee = name

    if details["rating"]>8:
       above_8.append(name)

sorted_names = sorted(employee, key=lambda name: employee[name]["rating"], reverse=True)
avg_rate=total_rating/n
print("Highest Rating:", highest_rating, ",Employee:", highest_rate_employee)
print("Average Rating:", avg_rate)
print("Employee with rating above 8: ",above_8)
print("Sorted employees by rating:",sorted_names)
```

- 15. Use dictionary comprehension to create an annual salary dictionary.

```python
annual_salary = {name: details["salary"] * 12 for name, details in employee.items()}
print("ANNUAL SALARY: ",annual_salary)
```

16. Use dictionary comprehension to create a dictionary of employees whose salary is above a selected threshold.

```python
threshold_salary = {name: details for name, details in employee.items() if details["salary"]>30000 }
print("Employees whose salary is above a selected threshold:")
for name,detail in threshold_salary.items():
    print("\t",name,":",detail)
```
