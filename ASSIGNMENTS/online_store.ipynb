```python
import pandas as pd
```

```python
prices = pd.Series([55000, 25000, 2500, 1800, 15000, 900])

print("PRODUCT PRICES")
print(prices)
```

```python
product_prices = pd.Series(
    [55000, 25000, 2500, 1800, 15000, 900],
    index=["Laptop", "Smartphone", "Headphones",
           "Keyboard", "Monitor", "Mouse"]
)

print("\nPRODUCT PRICES WITH CUSTOM INDEX")
print(product_prices)
```

```python
print("\nSERIES ACCESS")

print("Third product price:", prices.iloc[2])

print("Laptop price:", product_prices["Laptop"])

print("\nFirst three product prices:")
print(prices.iloc[0:3])
```

```python
data = {
    "Product_ID": ["P101", "P102", "P103", "P104", "P105", "P106"],
    "Product": [
        "Laptop", "Smartphone", "Headphones",
        "Keyboard", "Monitor", "Mouse"
    ],
    "Category": [
        "Electronics", "Electronics", "Accessories",
        "Accessories", "Electronics", "Accessories"
    ],
    "Price": [55000, 25000, 2500, 1800, 15000, 900],
    "Quantity": [3, 5, 10, 8, 4, 15],
    "Rating": [4.5, 4.2, 4.0, 4.3, 4.6, 4.1]
}

df = pd.DataFrame(data)

print(df)
```

```python
df
```

```python
print("\nSINGLE COLUMN - PRICE")
print(df["Price"])

print("\nMULTIPLE COLUMNS")
print(df[["Product", "Category", "Price"]])
```

```python
print("\nPRODUCTS WITH PRICE > 10000")
print(df[df["Price"] > 10000])

print("\nPRODUCTS WITH RATING > 4.3")
print(df[df["Rating"] > 4.3])

print("\nPRODUCTS WITH QUANTITY > 5")
print(df[df["Quantity"] > 5])

print("\nELECTRONICS PRODUCTS")
print(df[df["Category"] == "Electronics"])
```

```python
print("\nSPECIFIC CELL")
print("First product:", df.at[0, "Product"])
print("First product price:", df.at[0, "Price"])
```

```python
df["Sales_Amount"] = df["Price"] * df["Quantity"]

print("\nDATA WITH SALES AMOUNT")
print(df)
```

```python
total_sales = df["Sales_Amount"].sum()
average_price = df["Price"].mean()
highest_price = df["Price"].max()
lowest_price = df["Price"].min()
total_quantity = df["Quantity"].sum()

print("\nSALES STATISTICS")
print("Total Sales:", total_sales)
print("Average Price:", round(average_price, 2))
print("Highest Price:", highest_price)
print("Lowest Price:", lowest_price)
print("Total Quantity Sold:", total_quantity)
```

```python
best_selling_index = df["Quantity"].idxmax()
best_selling_product = df.loc[best_selling_index, "Product"]

highest_rated_index = df["Rating"].idxmax()
highest_rated_product = df.loc[highest_rated_index, "Product"]

print("\nBEST-SELLING PRODUCT")
print(best_selling_product)

print("\nHIGHEST-RATED PRODUCT")
print(highest_rated_product)
```

```python
category_sales = df.groupby("Category")["Sales_Amount"].sum()

print("\nSALES BY CATEGORY")
print(category_sales)
```
