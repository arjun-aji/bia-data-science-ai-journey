```python
import numpy as np
```

```python
marks = np.array([[85, 90, 88],
                  [78, 82, 80],
                  [92, 88, 95],
                  [80, 75, 85],
                  [95, 90, 92]])
```

```python
print("\nARRAY INFORMATION")
print("Number of dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Data type:", marks.dtype)
print("Total number of elements:", marks.size)
```

```python
print("\nINDEXING")
print("Python mark of Student 1:", marks[0, 0])
print("Database mark of Student 3:", marks[2, 1])
print("Web Development mark of Student 5:", marks[4, 2])
```

```python
# First 3 students
print("First 3 students:")
print(marks[:3])
```

```python
# Python marks of all students
print("\nPython marks:")
print(marks[:, 0])
```

```python
print("Marks + 5:")
print(marks + 5)
```

```python
print("\nMarks - 5:")
print(marks - 5)
```

```python
print("\nMarks * 2:")
print(marks * 2)
```

```python
student_average = np.mean(marks, axis=1)

for i in range(len(student_average)):
    print("Student", i + 1, "=", round(student_average[i], 2))
```

```python
subject_average = np.mean(marks, axis=0)

print("\nSUBJECT-WISE AVERAGES")
print("Python:", round(subject_average[0], 2))
print("Database:", round(subject_average[1], 2))
print("Web Development:", round(subject_average[2], 2))
```

```python
highest = np.max(marks)
lowest = np.min(marks)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
```

```python
standard_deviation = np.std(marks)
print("Standard deviation:", round(standard_deviation, 2))
```

```python
print("MARKS GREATER THAN 80")
print(marks[marks > 80])
```

```python
random_array = np.random.rand(3, 3)
print(random_array)
```

```python
total_marks = np.sum(marks)
average_marks = np.mean(marks)
print("     FINAL PERFORMANCE SUMMARY\n")
print("Total marks:", total_marks)
print("Overall average:", round(average_marks, 2))
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
print("Standard deviation:", round(np.std(marks), 2))

print("\nSubject-wise average:")
print("Python:", round(subject_average[0], 2))
print("Database:", round(subject_average[1], 2))
print("Web Development:", round(subject_average[2], 2))

print("\nStudent-wise average:")
for i, avg in enumerate(student_average):
    print("Student", i + 1, ":", round(avg, 2))

print("\nMarks greater than 80:")
print(marks[marks > 80])
```
