```python
Person_Name=input("1 . Enter Person Name: ")
Email_Address=input("2 . Enter Email Address: ")
Phone_Number=input("3 . Enter Phone Number: ")
Item_Name=input("4 . Enter Item Name: ")
Item_Description=input("5 . Enter Item Description: ")
Found_Location=input("6 . Enter Found Location: ")
Item_Category=input("7 . Enter Item Category: ")
Item_ID=input("8 . Enter Item ID: ")
Identifying_Mark=input("9 . Enter Identifying Mark: ")
Contact_Address=input("10. Enter Contact Address: ")

Person_Name=Person_Name.strip().title()
Item_Name=Item_Name.strip().capitalize()
Item_Category=Item_Category.strip().upper()
Found_Location=Found_Location.strip().title()
Email_Address=Email_Address.strip().lower()
Identifying_Mark=Identifying_Mark.strip().swapcase()
Item_Description=Item_Description.strip().capitalize()
Contact_Address=Contact_Address.strip().upper()

length=len(Phone_Number)
first_char=Person_Name[0]
second_char=Person_Name[1]
last_char=Person_Name[-1]
first_n=Person_Name[0:5]
last_n=Person_Name[-3:]
reverse=Person_Name[::-1]

search_word=Person_Name.split()[0]

find_result=Person_Name.find(search_word)
rfind_result=Person_Name.rfind(search_word)
index_result=Person_Name.index(search_word)
rindex_result=Person_Name.rindex(search_word)

in_result=search_word in Person_Name
not_in_result="XYZ" not in Person_Name

name_alpha=Person_Name.replace(" ","").isalpha()
phone_digit=Phone_Number.isdigit()
item_id_alnum=Item_ID.isalnum()
item_id_numeric=Item_ID.isnumeric()
item_id_decimal=Item_ID.isdecimal()
email_lower=Email_Address.islower()
name_title=Person_Name.istitle()

extra_space="   "+Person_Name+"   "
strip_result=extra_space.strip()
lstrip_result=extra_space.lstrip()
rstrip_result=extra_space.rstrip()

replace_result=Item_Description.replace("Black","Red")

email_split=Email_Address.split("@")
address_split=Contact_Address.split(",")
description_split=Item_Description.split()

email_join="@".join(email_split)
address_join=" - ".join(address_split)
description_join=" ".join(description_split)

prefix_result=Email_Address.startswith(search_word.lower())
suffix_result=Email_Address.endswith(".com")

name_count=Person_Name.count(search_word)
space_count=Person_Name.count(" ")

partition_result=Email_Address.partition("@")
rpartition_result=Email_Address.rpartition("@")

remove_prefix_result=Email_Address.removeprefix("www.")
remove_suffix_result=Email_Address.removesuffix(".com")

message="Lost item reported by "+Person_Name+" for the item "+Item_Name

report_message=f"{Person_Name} reported a {Item_Category.lower()} item named {Item_Name}."

center_name=Person_Name.center(30)
left_name=Person_Name.ljust(30)
right_name=Person_Name.rjust(30)

table=str.maketrans("aeiou","12345")
translated_name=Person_Name.lower().translate(table)

print("\n\n\t INFORMATION \n")

print("Person Name                  : ",Person_Name)
print("Email Address                : ",Email_Address)
print("Phone Number                 : ",Phone_Number)
print("Item Name                    : ",Item_Name)
print("Item Description             : ",Item_Description)
print("Found Location               : ",Found_Location)
print("Item Category                : ",Item_Category)
print("Item ID                      : ",Item_ID)
print("Identifying Mark             : ",Identifying_Mark)
print("Contact Address              : ",Contact_Address)

print("\n\t INDEXING & REVERSING \n")
print("Frist letter of person name  : ",first_char)
print("Second letter of person name : ",second_char)
print("Last letter of person name   : ",last_char)
print("Length of ph no              : ",length)
print("First 5 char of name         : ",first_n)
print("Last 5 char of name          : ",last_n)
print("Reversed name                : ",reverse)

print("\n\t SEARCH RESULTS \n")

print("find() result                : ",find_result)
print("rfind() result               : ",rfind_result)
print("index() result               : ",index_result)
print("rindex() result              : ",rindex_result)

print("\n\t MEMBERSHIP RESULTS \n")

print("Word present                 : ",in_result)
print("XYZ not present              : ",not_in_result)

print("\n\t VALIDATION RESULTS \n")

print("Name is alphabetic           : ",name_alpha)
print("Phone is digit               : ",phone_digit)
print("Item ID is alphanumeric      : ",item_id_alnum)
print("Item ID is numeric           : ",item_id_numeric)
print("Item ID is decimal           : ",item_id_decimal)
print("Email is lowercase           : ",email_lower)
print("Name is title case           : ",name_title)

print("\n\t STRIP RESULTS \n")

print("strip()                      : ",strip_result)
print("lstrip()                     : ",lstrip_result)
print("rstrip()                     : ",rstrip_result)

print("\n\t REPLACE RESULT \n")

print("Replaced description         : ",replace_result)

print("\n\t SPLIT RESULTS \n")

print("Email split                  : ",email_split)
print("Address split                : ",address_split)
print("Description split            : ",description_split)

print("\n\t JOIN RESULTS \n")

print("Joined email                 : ",email_join)
print("Joined address               : ",address_join)
print("Joined description           : ",description_join)

print("\n\t PREFIX & SUFFIX RESULTS \n")

print("Starts with search word      : ",prefix_result)
print("Ends with .com               : ",suffix_result)

print("\n\t COUNT RESULTS \n")

print("Search word count            : ",name_count)
print("Space count                  : ",space_count)

print("\n\t PARTITION RESULTS \n")

print("partition()                  : ",partition_result)
print("rpartition()                 : ",rpartition_result)

print("\n\t REMOVE PREFIX / SUFFIX \n")

print("removeprefix()               : ",remove_prefix_result)
print("removesuffix()               : ",remove_suffix_result)

print("\n\t MESSAGE \n")

print(message)

print("\n\t FORMATTED REPORT \n")

print(report_message)

print("\n\t ALIGNMENT \n")

print("Center : ",center_name)
print("Left   : ",left_name)
print("Right  : ",right_name)

print("\n\t TRANSLATE \n")

print("Translated name              : ",translated_name)
```
