### IF-ELIF-ELSE

```python
age = 70
if age<=18:
    print("child")
elif age<=30:
    print("adult")
elif age<=50:
    print("mid")
else :
    print("old")
```

```python
age = 70
gender = "m"
if age<=18:
    print("child")
    if gender == "m" :
        print("male")
    else :
        print("female")
elif age<=30:
    print("adult")
    if gender == "m" :
        print("male")
    else :
        print("female")
elif age<=50:
    print("mid")
    if gender == "m" :
        print("male")
    else :
        print("female")
else :
    print("old")
    if gender == "m" :
        print("male")
    else :
        print("female")
```

### break

```python
for i in range(1,11):
    print(i)
    if i==5:
        break
    print("after break")
```

### continue

```python
for i in range(1,11):
    print(i)
    if i==5:
        continue
    print("after continue")
```

```python
def add(*args):
    result=0
    for num in args:
        result+=num
    return result
```

```python
add(1,2)
```

```python
add(1,2,3,4)
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else :
        return n* fact(n-1)
    
```

```python
fact(5)
```

```python
fact(3)
```
