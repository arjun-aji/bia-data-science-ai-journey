```python
employee={"name":["ARJUN","DHRON","FEBIN"],"salary":[8000,6000,2000]}
highest_salary=max(employee["salary"])
print("HIGHEST_SALARY= ",highest_salary)
avg_salary=sum(employee["salary"])/len(employee["salary"])
print("AVERAGE_SALARY= ",avg_salary)
print("\nSALARY ABOVE 5000 : ")
for sal in employee["salary"]:
    if sal>5000:
        index= employee["salary"].index(sal)
        print("\t",employee["name"][index],", SALARY= ", sal)

```

```python
square_dict={x:x**2 for x in range(1,11) if x%2==0}
```

```python
square_dict
```

```python
{name:salary for name,salary in zip(employee["name"],employee["salary"]) if salary>5000}
```
