```python
customer_name= input("Enter customer name: ")         
mobile_brand = input("Enter mobile brand: ")            
mobile_model = input("Enter mobile model: ")             
mobile_price = float(input("Enter mobile price: "))
quantity  = int(input("Enter quantity: "))
accessories_name= input("Enter accesories name separated ny comma : ")
accessories_cost = float(input("Enter accessories cost: "))
warranty_cost = float(input("Enter warranty cost: "))
discount_percentage = float(input("Enter discount percentage: "))
tax_percentage = float(input("Enter tax percentage: "))
advance= float(input("Enter advance payment: "))
emi_months= int(input("No of EMI months: "))

customer_name = customer_name.strip().title()      
mobile_brand  = mobile_brand.strip().upper()        
mobile_model  = mobile_model.strip().capitalize()   

full_product_name = mobile_brand + " " + mobile_model   

total_mobile_cost = mobile_price * quantity              

subtotal = total_mobile_cost + accessories_cost + warranty_cost   

discount_amount = subtotal * discount_percentage / 100       

amount_after_discount = subtotal-discount_amount               

tax_amount = amount_after_discount * tax_percentage / 100

final_payable = amount_after_discount+tax_amount
pending_amount = final_payable-advance
                    
is_advance_valid = advance <= final_payable and advance >= 0
is_full_payment = advance == final_payable
discount_or_tax_applied = discount_percentage > 0 or tax_percentage > 0

monthly_emi = pending_amount/ emi_months

known_brands = ["APPLE", "SAMSUNG", "ONEPLUS" , "VIVO", "OPPO", "REDMI", "REALME"]

is_known_brand = mobile_brand in known_brands           
is_unknown_brand = mobile_brand not in known_brands       

price_is_float = type(mobile_price) is float           

print(f"{'\n\n\tCUSTOMER BILL\n'}")
print(f"Customer Name        : {customer_name}")
print(f"Mobile Brand         : {mobile_brand}")
print(f"Mobile Model         : {mobile_model}")
print(f"Full Product Name    : {full_product_name}")
print(f"Quantity             : {quantity}")
print(f"Accessories purchased: {accessories_name}")
print(f"Total Mobile Cost    : Rs {total_mobile_cost:.2f}")
print(f"Accessories Cost     : Rs {accessories_cost:.2f}")
print(f"Warranty Cost        : Rs {warranty_cost:.2f}")
print(f"Subtotal             : Rs {subtotal:.2f}")
print(f"Discount ({discount_percentage}%)     : Rs {discount_amount:.2f}")
print(f"Amount After Discount: Rs {amount_after_discount:.2f}")
print(f"Tax ({tax_percentage}%)           : Rs{tax_amount:.2f}")
print(f"Final Payable Amount : Rs {final_payable:.2f}")
print(f"Advance Payment      : Rs {advance:.2f}")
print(f"Remaining Amount     : Rs {pending_amount:.2f}")
print(f"EMI Plan             : {emi_months} months")
print(f"Monthly EMI          : Rs {monthly_emi:.2f}")
print("\n\tSYSTEM CHECKS\n")
print(f"Advance payment is valid                : {is_advance_valid}")
print(f"Advance payment equals final amount     : {is_full_payment}")
print(f"Discount or tax applied                 : {discount_or_tax_applied}")
print(f"Brand '{mobile_brand}' is a known brand          : {is_known_brand}")
print(f"Brand '{mobile_brand}' is an unrecognized brand  : {is_unknown_brand}")
print(f"Price stored correctly as float         : {price_is_float}")
print("\n\tThank you for shopping with us!")
   
```
