```python
players = ["Arun", "Rahul", "Vishnu", "Akash", "Manu"]

print("PLAYER LIST")
print(players)
```

```python
players.append("Akhil")
print("After append:", players)
```

```python
players.extend(["Rohan", "Nikhil"])
print("After extend:", players)
```

```python
players.insert(2, "Sree")
print("After insert:", players)
```

```python
players.remove("Rahul")
print("After remove:", players)
```

```python
removed_player = players.pop(1)
print("Removed player:", removed_player)
print("After pop:", players)
```

```python
del players[1]
print("After del:", players)
```

```python
print("Is Arun in the team?", "Arun" in players)
print("Number of players:", len(players))
```

```python
players.reverse()
print("Reversed list:", players)
```

```python
print("Count of Arun:", players.count("Arun"))
```

```python
copied_players = players.copy()
print("Copied list:", copied_players)
```

```python
print("First player:", players[0])
print("Third player:", players[2])
print("Last player:", players[-1])
print("Second-last player:", players[-2])
```

```python
group1 = players[1:4]
group2 = players[::2]
group3 = players[-3:]
group4 = players[::-1]

print("Selected range:", group1)
print("Every second player:", group2)
print("Last few players:", group3)
print("Reversed group:", group4)
```

```python
players[1:3] = ["Thomas", "Joel"]
print("After slice assignment:", players)
```

```python
selected_players = [name for name in players if len(name) > 4]
name_lengths = [len(name) for name in players]

print("Players with more than 4 letters:", selected_players)
print("Length of player names:", name_lengths)
```

```python
match = (
    "M001",
    "College Tournament",
    "ABC College",
    "College Ground",
    "20-09-2026"
)

print("MATCH INFORMATION")
print(match)

print("Match ID:", match[0])
print("Last item:", match[-1])
print("Match details:", match[1:4])


match_id, tournament, opponent, venue, date = match

print("Match ID:", match_id)
print("Tournament:", tournament)
print("Opponent:", opponent)
print("Venue:", venue)
print("Date:", date)


match1 = ("M001", "ABC College")
match2 = ("20-09-2026", "College Ground")

combined = match1 + match2
repeated = match1 * 2

print("Combined tuple:", combined)
print("Repeated tuple:", repeated)


players_tuple = tuple(players)
print("Players as tuple:", players_tuple)

players_list = list(players_tuple)
print("Tuple converted to list:", players_list)
```
