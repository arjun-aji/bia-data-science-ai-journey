```python
number_of_products= input("Enter the number of products")
number_of_products= int(number_of_products)
product_count =0
bill_amount=0
highest_price=0
highest_product=""
lowest_price=0
lowest_product=""

for i in range(number_of_products):
    product_name=input("Enter product name: ")
    product_price=input("Enter product price: ")
    product_price=int(product_price)
    while True:
        product_quantity=input("Enter product quantity: ")
        product_quantity=int(product_quantity)
        if product_quantity <= 0:
                print("Quantity must be greater than zero")
                continue
        break
    product_total=product_price*product_quantity
    print(f"{product_name} Total= {product_total} ")
    bill_amount+=product_total
    product_count+=1

    if product_count==1:
        highest_price=product_price
        highest_product=product_name
        lowest_price=product_price
        lowest_product=product_name

    else:
        if product_price>highest_price:
            highest_price=product_price
            highest_product=product_name
            
        if product_price<lowest_price:
            lowest_price=product_price
            lowest_product=product_name

if bill_amount>2000:
    discount_rate=10
elif bill_amount>1000:
    discount_rate=5
else:
    discount_rate=0
discount_amount= bill_amount*discount_rate/100
final_bill_amount= bill_amount-discount_amount

print("======INVOINCE======")
print(f"No of product= {product_count}")
print(f"Total amount= Rs {bill_amount}/-")
print(f"DISCOUNT RATE= {discount_rate}%")
print(f"DISCOUNT AMOUNT= Rs {discount_amount}/-")
print(f"FINAL AMOUNT= Rs {final_bill_amount}/-")
print("=========================")
print(f"HIGHEST PRODUCT= {highest_product}, PRICE= Rs {highest_price}/-")
print(f"LOWEST PRODUCT= {lowest_product}, PRICE= Rs {lowest_price}/-")
```
