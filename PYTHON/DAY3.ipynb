```python
employees=[("Arjun",500,3),("Varun",300,2),("Ajmal",400,1),("Dhron",200,4)]
highest_salary= 0
highest_sal_name=""
total_salary=0
for employee in employees:
    name,rate,hours=employee
    salary= rate*hours
    print("Employee Name= ",name," ," ,"SALARY= Rs" ,salary)
    total_salary+= salary
    if salary>highest_salary:
        highest_salary=salary
        highest_sal_name=name
print("\nHIGHEST SALARY= ",highest_salary," ," ,"Employee Name= ",highest_sal_name)
print("Average Salary= Rs",total_salary/len(employees))
    
```

```python
course_info=(("CST404",["ARJUN","DHRON"],[90,85]),("CSL401",["FEBIN","JAIJITH"],[65,80]))
for course in course_info:
    course_name,student,grade=course
    print("COURSE_ID= ",course_name,"\nSTUDENTS= ",student,"\nGRADES= ",grade," \nAVERAGE GRADE=", sum(grade)/len(grade))
    high= max(grade)
    print("TOP STUDENT= ",student[grade.index(high)],"\nHIGHEST GRADE =",high,"\n")
```

```python
name=input("enter student name to update: ")
mark=input("enter new grade: ")
mark=int(mark)
for course in course_info:
    course_name, student, grade = course
    if name in student:
        index = student.index(name)
        grade[index] = mark  
for course in course_info:
    course_name,student,grade=course
    print("COURSE_ID= ",course_name,"\nSTUDENTS= ",student,"\nGRADES= ",grade," \nAVERAGE GRADE=", sum(grade)/len(grade))
    high= max(grade)
    print("TOP STUDENT= ",student[grade.index(high)],"\nHIGHEST GRADE =",high,"\n")
```
